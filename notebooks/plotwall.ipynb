```python

import imas
from idstools.view.common import Canvas
from idstools.view.wall import WallView
```

Following is the wall information of different tokamaks

```python
WALL = {
    "ITER": ("ITER_MD", 116000, 2, "public"),
    "WEST": ("west", 54178, 0, "fleuryl"),
    "TCV": ("tcv", 64770, 1, "fleuryl"),
    "AUG": ("aug", 40866, 1, "costerd"),
    "JET": ("jet", 92436, 271, "schneim"),
    "JT60": ("JT-60U", 33654, 1, "schneim"),
    "D3D": ("D3D", 133221, 0, "smiths"),
}
```

Select Machine and get database details

```python
MACHINE = "ITER"
W = WALL[MACHINE]
```

Connect to database and retrive wall IDS

```python
db = imas.DBEntry(imas.imasdef.MDSPLUS_BACKEND, W[0], W[1], W[2], W[3])
db.open()
wall = db.get("wall")
```

Initialize canvas and add axes and set attributes

```python
# Initialization
canvas = Canvas(1, 1)
ax = canvas.add_axes(
    title="Machine Description", xlabel="R (m)", ylabel="Z (m)", row=0, col=0
)
canvas.fig.set_size_inches(7.5, 10.0)
ax.set_aspect("equal", adjustable="box")
```

Get Wall data fro plotting and plot on axes

```python
wallview = WallView(wall)
wallview.view_wall(ax)
```

Plot axes and show canvas

```python
ax.plot()

canvas.show()
```
